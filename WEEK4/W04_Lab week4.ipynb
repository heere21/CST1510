## Topic 4 · Handling errors
*Walkthrough: `04 - Handling Errors`*

### D14. Catch a `ValueError`

1. Create `text = "abc"`.
2. Inside `try:`, convert it with `float(text)` and print the number.
3. In `except ValueError:`, print `Not a number: abc` instead.
4. Now change `text` to `"42"` and run again.

**Expected output:** `Not a number: abc`, then (after the change) `42.0`

````python
# D14
"""Catch a ValueError"""

text = "abc" #set text variables to 'abc'
try: #start try block to test code for erros
    number = float(text) #try converting text to a float
    print(number) #Print the converted number if succesful
except ValueError: #Catch error if conversion fails
    print(f"Not a number: {42}") #print fallback message
````

### D15. Catch a `FileNotFoundError`

1. Inside `try:`, open `missing.txt` with `with` and print its contents.
2. In `except FileNotFoundError:`, print `missing.txt not found`.

**Expected output:** `missing.txt not found`

````python
# D15
"""Catch a FileNotFoundError"""
try:
    with open("missing.txt") as f: #Try to open the non-existent file
        print(f.read()) #print file contents if found

except FileNotFoundError: #Catch error when the file does not exist
    print("missing.txt not found") #print fallback message
````

### C1. Load records safely from a file

Write `load_records(path)` with a one-line docstring. It reads `label,value,limit` lines from a file
and returns a **list of dicts**. If the file does not exist, it prints a clear message and returns an
empty list instead of crashing.

Then load `records_sample.txt` and print each record's label and status, as in D16.

**Test it:**

| Call | Expected output |
|---|---|
| `load_records("records_sample.txt")` | the same three lines as D16 |
| `load_records("nope.txt")` | a clear message, then `[]` |

````python
# C1
def load_records(path):
    """load records from a file into a list of dictionaries"""
    records = [] #Initialise an empty list to store records
    try: #try to open and read file
        with open(path) as f: #Open specified file path
            for line in f: #Loop through each line in the file
                label, value, limit = (line.strip().split(","))
                record = {
                    "label": label,
                    "value": float(value),
                    "limit": float(limit)
                }
                records.append(record) #Add dictionary to the list
    except FileNotFoundError:
        print(f"{path} not found")
    return records #Return the list of dicts or an empty list 
   
    records = load_records("records_sample.txt")
    for r in records:
        print(r)
    print(load_records("nope.txt"))
````

### C2. Write the results to a file

Using your records from C1, write one line per record to a new file called `results.txt`,
in the form `label,status`. Then read `results.txt` back and print it.

**Expected output:**
```
R-001,OK
R-002,OVER LIMIT
R-003,OK
```

````python
# C2
records = load_records("records_sample.txt")
#Open new file to write the results
with open("results.txt", "w") as f:
    #Loop through each record from C1
    for r in records:
        #Check if the value is within the limit
        if r["value"] <= r["limit"]:
            status = "OK"
        else:
            status = "OVER LIMIT"
        f.write(f"{r['label']},{status}\n")
#Open file again to read and print the contents
with open("results.txt", "r") as f:
    print(f.read())
````
